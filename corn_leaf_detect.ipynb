# Corn Leaf Disease Detection with OpenCV
Load the trained `corn_leaf_model.h5`.

```python
import cv2
```

```python
from tensorflow.keras.models import load_model
import numpy as np
```

## Load the model

```python
model = load_model('corn_leaf_model.h5', compile=False)   # compile=False hides the optimizer warnings; we only need predictions
class_names = ['diseased', 'healthy']                      # same order the model was trained with (index 0, index 1)
```

## Prediction function

```python
def leaf_detector(frame):
    leaf = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)   
    leaf = cv2.resize(leaf, (224, 224))
    leaf = leaf.reshape(1, 224, 224, 3)

    pred = model.predict(leaf)[0][0]
    print(pred)

    if pred > 0.5:
        label = 'HEALTHY'
        confidence = pred
        color = (255, 255, 255)        # white (BGR)
    else:
        label = 'DISEASED'
        confidence = 1 - pred
        color = (0, 0, 255)        # red (BGR)

    text = f'{label}'

    # scale the text to the image size so it is readable on any photo
    scale = max(frame.shape[1] / 600, 0.6)
    thickness = max(int(scale * 2), 1)
    cv2.putText(frame, text, (10, int(40 * scale)),
                cv2.FONT_HERSHEY_COMPLEX, scale, color, thickness)

    return frame, label
```

## Test on a photo

```python
img1 = cv2.imread('Healthy1.jpg')      
result, label = leaf_detector(img1)
print('Result:', label)
cv2.imshow('image', result)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
img2 = cv2.imread('Diseased1.jpg')     
result, label = leaf_detector(img2)
print('Result:', label)
cv2.imshow('image', result)
cv2.waitKey()
cv2.destroyAllWindows()
```

## Webcam (optional)
Hold a leaf in front of the camera. Press **q** to quit.

```python
video = cv2.VideoCapture(0)
while True:
    suc, image = video.read()
    if not suc:
        break
    result, label = leaf_detector(image)
    cv2.imshow('image', result)
    if cv2.waitKey(1) & 0XFF == ord('q'):
        break
video.release()
cv2.destroyAllWindows()
```
